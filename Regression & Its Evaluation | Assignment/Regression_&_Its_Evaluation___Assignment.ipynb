# Question 1: What is Simple Linear Regression?

**Simple Linear Regression** is a supervised Machine Learning/statistical technique used to model the relationship between **one independent variable (X)** and **one dependent variable (Y)**.

It assumes that the relationship between X and Y can be represented by a straight line.

### Formula

$$
Y = \beta_0 + \beta_1X + \epsilon
$$

Where:

* \(Y\) = dependent/target variable
* \(X\) = independent/input variable
* \(\beta_0\) = intercept
* \(\beta_1\) = slope/coefficient
* \(\epsilon\) = error term

The predicted value is commonly written as:

$$
\hat{Y} = \beta_0 + \beta_1X
$$

### Example

Suppose we want to predict a student's marks based on study hours.

```text
Study Hours → Marks
     1      →  40
     2      →  50
     3      →  60
```

Here:

* Study Hours = independent variable
* Marks = dependent variable

### Applications

* Predicting house prices
* Predicting sales
* Predicting salary
* Predicting demand
* Predicting marks

### Easy Definition

**Simple Linear Regression predicts a dependent variable using one independent variable through a straight-line relationship.**

# Question 2: What Are the Key Assumptions of Simple Linear Regression?

Simple Linear Regression relies on several important assumptions.

## 1. Linearity

There should be a reasonably linear relationship between X and Y.

```text
X ↑ → Y tends to change in a straight-line pattern
```

## 2. Independence

Observations/errors should be independent of each other.

One observation should not systematically depend on another.

## 3. Homoscedasticity

The variance of the residuals should remain approximately constant across predicted values.

If residual spread changes significantly, this is called **heteroscedasticity**.

## 4. Normality of Residuals

For classical statistical inference, residuals are generally assumed to be approximately normally distributed.

This assumption is especially important for reliable confidence intervals and hypothesis tests.

## 5. No Significant Outliers/Influential Points

Extreme observations should not disproportionately control the regression line.

## 6. Correct Model Specification

The model should include appropriate predictors and functional form.

### Summary Table

| Assumption              | Meaning                                 |
| ----------------------- | --------------------------------------- |
| Linearity               | X and Y have a linear relationship      |
| Independence            | Observations/errors are independent     |
| Homoscedasticity        | Constant residual variance              |
| Normality               | Residuals are approximately normal      |
| No influential outliers | Extreme points should not dominate      |
| Correct specification   | Appropriate variables/form are included |

### Easy Definition

**Good linear regression requires linearity, independence, constant error variance, suitable residual behavior, and no problematic influential observations.**

# Question 3: What Is Heteroscedasticity, and Why Is It Important to Address in Regression Models?

**Heteroscedasticity** occurs when the variance of regression errors/residuals is **not constant** across the range of fitted values or an explanatory variable.

In simple terms, the residuals may form a pattern where their spread increases or decreases.

### Example

Suppose we predict house prices.

For inexpensive houses:

```text
Small prediction errors
```

For expensive houses:

```text
Large prediction errors
```

The changing spread of errors indicates heteroscedasticity.

### Visual Representation

A residual plot may look like:

```text
Residual
   |
   |       .
   |      . .
   |    .   . .
   |  .       .  .
   |________________ Predicted Value
```

This widening pattern is often called a **funnel shape**.

### Why Is It Important?

Heteroscedasticity can:

* Make standard errors unreliable.
* Affect confidence intervals.
* Affect hypothesis-test results.
* Make coefficient significance tests less trustworthy.
* Indicate that the model does not capture the error structure well.

The ordinary least-squares coefficient estimates can still be unbiased under some conditions, but their usual standard errors may be incorrect.

### How Can It Be Detected?

Residual plots can be used.

```python
plt.scatter(y_pred, residuals)
```

Formal tests such as the **Breusch-Pagan test** can also be used.

### Ways to Handle It

1. Transform the target, e.g. log transformation.
2. Transform relevant predictors.
3. Use weighted least squares.
4. Use heteroscedasticity-robust standard errors.
5. Investigate missing variables or incorrect model specification.

### Easy Definition

**Heteroscedasticity = Unequal variance of regression errors.**

# Question 4: What is Multiple Linear Regression?

**Multiple Linear Regression** is a regression technique used to predict a dependent variable using **two or more independent variables**.

### Formula

$$
Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + \cdots + \beta_kX_k + \epsilon
$$

Where:

* \(Y\) = dependent variable
* \(X_1, X_2, ..., X_k\) = independent variables
* \(\beta_0\) = intercept
* \(\beta_1, \beta_2, ..., \beta_k\) = coefficients
* \(\epsilon\) = error

### Example

House price can depend on:

* Area
* Number of rooms
* Age of house
* Location

Therefore:

$$
Price = \beta_0 + \beta_1Area + \beta_2Rooms + \beta_3Age + \epsilon
$$

### Python Example

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [
    [1000, 2],
    [1500, 3],
    [2000, 4]
]

y = [200000, 300000, 400000]

model.fit(X, y)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

### Difference from Simple Linear Regression

| Simple Linear Regression | Multiple Linear Regression      |
| ------------------------ | ------------------------------- |
| One predictor            | Two or more predictors          |
| \(Y=\beta_0+\beta_1X\)   | \(Y=\beta_0+\beta_1X_1+\cdots\) |
| Simpler                  | More complex                    |
| One feature              | Multiple features               |

### Easy Definition

**Multiple Linear Regression predicts one target using two or more independent variables.**

# Question 5: What Is Polynomial Regression, and How Does It Differ from Linear Regression?

**Polynomial Regression** is used when the relationship between the independent and dependent variables is better represented by a curved relationship rather than a straight line.

### Formula

For a second-degree polynomial:

$$
Y = \beta_0 + \beta_1X + \beta_2X^2 + \epsilon
$$

For a third-degree polynomial:

$$
Y = \beta_0 + \beta_1X + \beta_2X^2 + \beta_3X^3 + \epsilon
$$

### Example

```text
Linear Regression:

Y
|       /
|     /
|   /
| /
|____________ X


Polynomial Regression:

Y
|        .
|      .
|    .
|  .
|.____________ X
```

### Important Point

Polynomial Regression is **linear in its coefficients**, even though it models a nonlinear relationship in X.

For example:

$$
Y = \beta_0+\beta_1X+\beta_2X^2
$$

is still fitted using linear regression after creating polynomial features.

### Python Example

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

model = LinearRegression()

model.fit(X_poly, y)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

### Difference

| Linear Regression                          | Polynomial Regression   |
| ------------------------------------------ | ----------------------- |
| Straight-line relationship                 | Curved relationship     |
| Uses X                                     | Uses X, X², X³, etc.    |
| Simpler                                    | More flexible           |
| Less risk of overfitting at low complexity | High degree can overfit |

### Easy Definition

**Polynomial Regression extends linear regression by adding powers of the features to model curved relationships.**

# Question 6: Implement Simple Linear Regression

### Given Data

```text
X = [1, 2, 3, 4, 5]

Y = [2.1, 4.3, 6.1, 7.9, 10.2]
```

## Python Program

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

# Input data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])

# Create model
model = LinearRegression()

# Train model
model.fit(X, Y)

# Predictions
Y_pred = model.predict(X)

# Display model parameters
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)

# Predict a new value
new_prediction = model.predict([[6]])
print("Prediction for X = 6:", new_prediction[0])

# Plot actual data
plt.figure(figsize=(8, 5))

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, Y_pred, linewidth=2, label="Regression Line")

plt.title("Simple Linear Regression")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()

plt.show()
```

## Expected Model

The fitted line will be approximately:

$$
\hat{Y} \approx 0.08 + 2.00X
$$

The exact coefficient values can be obtained by running the code.

### Interpretation

The slope is approximately **2**, meaning that an increase of one unit in X is associated with an increase of about 2 units in predicted Y.

# Question 9: Create a Residuals Plot for a Regression Model

### Given Data

```text
X = [10, 20, 30, 40, 50]

Y = [15, 35, 40, 50, 65]
```

## Python Program

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

# Data
X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
Y = np.array([15, 35, 40, 50, 65])

# Create and train model
model = LinearRegression()

model.fit(X, Y)

# Predictions
Y_pred = model.predict(X)

# Calculate residuals
residuals = Y - Y_pred

print("Actual Values:", Y)
print("Predicted Values:", Y_pred)
print("Residuals:", residuals)

# Residual plot
plt.figure(figsize=(8, 5))

plt.scatter(Y_pred, residuals)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=2
)

plt.title("Residuals Plot")
plt.xlabel("Predicted Values")
plt.ylabel("Residuals")

plt.show()
```

## What Is a Residual?

A residual is the difference between the actual value and predicted value.

$$
Residual = Y-\hat{Y}
$$

For example, if:

```text
Actual = 50
Predicted = 47
```

then:

$$
Residual=50-47=3
$$

### How to Interpret the Residual Plot

A good residual plot generally shows:

* Points randomly scattered around zero.
* No obvious curve.
* No obvious funnel/widening pattern.
* Approximately constant spread.

### If We See a Pattern

A pattern may indicate:

* Non-linearity
* Heteroscedasticity
* Missing variables
* Incorrect model specification

### Easy Definition

**Residual = Actual value − Predicted value.**
