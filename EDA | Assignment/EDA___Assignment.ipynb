### Visualize Missing Values

```python id="missingplot"
plt.figure(figsize=(10, 5))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

## Approach for Handling Missing Values

The approach depends on the column and the percentage of missing values.

### Numerical Columns

For numerical columns such as `selling_price` or `km_driven`, we can use the **median** when appropriate:

```python id="medianfill"
df["km_driven"] = df["km_driven"].fillna(df["km_driven"].median())
```

Median is often preferred when the data contains outliers because it is less affected by extreme values.

### Categorical Columns

For categorical columns such as `seller_type` or `owner`, we can use the **mode**:

```python id="modefill"
df["seller_type"] = df["seller_type"].fillna(df["seller_type"].mode()[0])
```

### If Missing Values Are Very Few

Rows can sometimes be removed:

```python id="dropmissing"
df = df.dropna()
```

### Conclusion

**My approach:** First identify the number and percentage of missing values. Then use median/mode imputation where appropriate, and remove rows only when the amount of missing data is small or the records are not useful.

# Question 6: Use the IQR Method to Detect and Remove Outliers from km_driven. Show Before-and-After Summary Statistics

## Step 1: Before Cleaning

```python id="beforeiqr"
print("Before Removing Outliers:")
print(df["km_driven"].describe())
```

## Step 2: Calculate IQR

The **Interquartile Range (IQR)** is:

$$
IQR = Q3 - Q1
$$

where:

* Q1 = 25th percentile
* Q3 = 75th percentile

The lower and upper limits are:

$$
Lower = Q1 - 1.5 \times IQR
$$

$$
Upper = Q3 + 1.5 \times IQR
$$

### Python Code

```python id="iqr"
Q1 = df["km_driven"].quantile(0.25)
Q3 = df["km_driven"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

## Step 3: Detect Outliers

```python id="detectout"
outliers = df[
    (df["km_driven"] < lower_bound) |
    (df["km_driven"] > upper_bound)
]

print("Number of Outliers:", len(outliers))
display(outliers.head())
```

## Step 4: Remove Outliers

```python id="removeout"
df_clean = df[
    (df["km_driven"] >= lower_bound) &
    (df["km_driven"] <= upper_bound)
].copy()
```

## Step 5: After Cleaning

```python id="afteriqr"
print("After Removing Outliers:")
print(df_clean["km_driven"].describe())
```

## Before vs After Summary

```python id="compareiqr"
before = df["km_driven"].describe()
after = df_clean["km_driven"].describe()

comparison = pd.DataFrame({
    "Before": before,
    "After": after
})

display(comparison)
```

## Observation

After applying the IQR method, extreme values outside the lower and upper bounds are removed.

The **maximum km_driven and overall spread should decrease** if high-distance outliers were present.

# Question 8: Convert seller_type into Numeric Format Using One-Hot Encoding

One-hot encoding converts categorical values into separate binary columns.

For example:

```text
seller_type
-----------
Individual
Dealer
```

can become:

```text
seller_type_Dealer
seller_type_Individual
```

## Python Code
