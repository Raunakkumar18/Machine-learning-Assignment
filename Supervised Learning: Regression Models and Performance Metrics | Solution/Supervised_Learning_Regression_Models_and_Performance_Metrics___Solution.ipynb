## Question 1: What is Simple Linear Regression (SLR)? Explain Its Purpose.

**Simple Linear Regression (SLR)** is a statistical and Machine Learning technique used to study and model the relationship between **one independent variable (X)** and **one dependent variable (Y)**.

It fits a straight line to the data and uses that line to estimate or predict the value of Y based on X.

### Purpose of SLR

The main purposes are:

1. To understand the relationship between two variables.
2. To predict the dependent variable.
3. To measure how changes in X are associated with changes in Y.
4. To estimate the strength and direction of the relationship.

### Example

If we want to predict **salary based on years of experience**:

```text
Years of Experience → Salary
```

Years of experience is X, and salary is Y.

### Easy Definition

**Simple Linear Regression predicts one dependent variable using one independent variable through a straight-line relationship.**

## Question 2: What Are the Key Assumptions of Simple Linear Regression?

The major assumptions are:

### 1. Linearity

There should be an approximately linear relationship between X and Y.

### 2. Independence

The observations and regression errors should be independent of each other.

### 3. Homoscedasticity

The variance of the residuals should remain approximately constant across the range of fitted values.

If the variance changes, it is called **heteroscedasticity**.

### 4. Normality of Residuals

For conventional statistical inference, the residuals are generally assumed to be approximately normally distributed.

### 5. No Problematic Outliers

Extreme or influential observations should not disproportionately affect the regression line.

### Summary

| Assumption              | Meaning                                      |
| ----------------------- | -------------------------------------------- |
| Linearity               | X and Y have a linear relationship           |
| Independence            | Observations/errors are independent          |
| Homoscedasticity        | Constant residual variance                   |
| Normality               | Residuals are approximately normal           |
| No influential outliers | Extreme points should not dominate the model |

### Easy Definition

**SLR assumes linearity, independence, constant error variance, approximately normal residuals for inference, and no problematic influential observations.**

## Question 3: Write the Mathematical Equation for a Simple Linear Regression Model and Explain Each Term.

The mathematical equation is:

$$
Y = \beta_0 + \beta_1X + \epsilon
$$

For prediction, we commonly write:

$$
\hat{Y} = \beta_0 + \beta_1X
$$

### Explanation of Terms

| Term         | Meaning                   |
| ------------ | ------------------------- |
| \(Y\)        | Actual dependent variable |
| \(\hat{Y}\)  | Predicted value of Y      |
| \(X\)        | Independent variable      |
| \(\beta_0\)  | Intercept                 |
| \(\beta_1\)  | Slope/coefficient         |
| \(\epsilon\) | Error term                |

### Intercept

$$
\beta_0
$$

is the expected value of Y when X is 0.

### Slope

$$
\beta_1
$$

represents the expected change in Y for a **one-unit increase in X**.

### Example

Suppose:

$$
\hat{Y}=10+5X
$$

Here:

* Intercept = 10
* Slope = 5

If X increases by 1, predicted Y increases by 5.

## Question 5: What Is the Method of Least Squares in Linear Regression?

The **Least Squares Method** is a technique used to find the best-fitting regression line by minimizing the **sum of squared residuals**.

A residual is:

$$
e_i = y_i-\hat{y}_i
$$

The least-squares method minimizes:

$$
\boxed{\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
$$

In simple linear regression:

$$
\hat{y}_i=\beta_0+\beta_1x_i
$$

Therefore, the method chooses the intercept and slope that make the total squared prediction error as small as possible.

### Why Square the Errors?

Squaring:

* Makes all errors positive.
* Gives greater weight to larger errors.
* Provides a convenient mathematical optimization objective.

### Example

If actual value = 100 and predicted value = 90:

$$
Error=100-90=10
$$

Squared error:

$$
10^2=100
$$

The regression algorithm tries to choose the line that minimizes the total squared error across all observations.

### Easy Definition

**Least Squares = Find the regression line that minimizes the sum of squared prediction errors.**

## Question 6: What Is Logistic Regression? How Does It Differ from Linear Regression?

**Logistic Regression** is a supervised Machine Learning algorithm primarily used for **classification**, especially binary classification.

It predicts the probability of an observation belonging to a class.

For binary classification, the logistic/sigmoid function maps a linear combination of inputs to a value between 0 and 1:

$$
p=\frac{1}{1+e^{-z}}
$$

where:

$$
z=\beta_0+\beta_1X
$$

### Example

Predict whether a customer will purchase a product:

```text
1 → Purchase
0 → No Purchase
```

### Logistic vs Linear Regression

| Linear Regression                 | Logistic Regression                    |
| --------------------------------- | -------------------------------------- |
| Used mainly for regression        | Used mainly for classification         |
| Predicts continuous values        | Predicts class probabilities           |
| Output can be any real number     | Probability is between 0 and 1         |
| Uses a linear prediction equation | Uses a logistic/sigmoid transformation |
| Example: predict salary           | Example: predict yes/no                |

### Example

Linear Regression:

```text
Predict house price = ₹5,00,000
```

Logistic Regression:

```text
Probability customer buys = 0.82
```

A classification threshold can then be applied, such as 0.5, to convert probability into a class label.

### Easy Definition

**Linear Regression predicts continuous values, while Logistic Regression predicts probabilities/classes.**

## Question 7: Name and Briefly Describe Three Common Evaluation Metrics for Regression Models.

Three common regression evaluation metrics are:

### 1. Mean Absolute Error (MAE)

MAE measures the average absolute difference between actual and predicted values.

$$
MAE=\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

Lower MAE indicates smaller average prediction errors.

---

### 2. Mean Squared Error (MSE)

MSE calculates the average squared prediction error.

$$
MSE=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Because errors are squared, large errors receive more weight.

Lower MSE is better.

---

### 3. Root Mean Squared Error (RMSE)

RMSE is the square root of MSE.

$$
RMSE=\sqrt{MSE}
$$

It is expressed in the same units as the target variable.

Lower RMSE indicates better predictive performance.

### Comparison

| Metric | Meaning                | Lower is better? |
| ------ | ---------------------- | ---------------- |
| MAE    | Average absolute error | Yes              |
| MSE    | Average squared error  | Yes              |
| RMSE   | Square root of MSE     | Yes              |

## Question 8: What Is the Purpose of the R-squared Metric in Regression Analysis?

**R-squared (\(R^2\))** measures the proportion of variation in the dependent variable that is explained by the regression model.

A common formula is:

$$
R^2=1-\frac{SS_{res}}{SS_{tot}}
$$

Where:

* \(SS_{res}\) = residual sum of squares
* \(SS_{tot}\) = total sum of squares

### Example

If:

$$
R^2=0.80
$$

then the model explains approximately **80% of the variation in the target variable within the evaluated dataset**.

### Interpretation

* \(R^2\) closer to 1 → model explains more variation.
* \(R^2\) closer to 0 → model explains less variation.

For models with an intercept, \(R^2\) is commonly between 0 and 1, but it can be negative on test data or in other settings when the model performs worse than the baseline mean predictor.

### Important Point

A high \(R^2\) does not automatically mean the model is good. We should also consider:

* MAE
* RMSE
* Residual plots
* Test-set performance
* Model assumptions
* Business context

### Easy Definition

**R² tells us how much of the variation in the target is explained by the regression model.**

## Question 9: Write Python Code to Fit a Simple Linear Regression Model Using Scikit-learn and Print the Slope and Intercept.

### Python / Google Colab Code

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Independent variable
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)

# Dependent variable
Y = np.array([2, 4, 6, 8, 10])

# Create the model
model = LinearRegression()

# Fit the model
model.fit(X, Y)

# Print slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

### Expected Output

```text
Slope: 2.0
Intercept: 0.0
```

Therefore, the regression equation is:

$$
\hat{Y}=0+2X
$$

or:

$$
\boxed{\hat{Y}=2X}
$$

### Explanation

```python
model.fit(X, Y)
```

trains the linear regression model.

```python
model.coef_[0]
```

returns the **slope**.

```python
model.intercept_
```

returns the **intercept**.

### Visualization

```python
import matplotlib.pyplot as plt

Y_pred = model.predict(X)

plt.scatter(X, Y, label="Actual Data")
plt.plot(X, Y_pred, linewidth=2, label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

## Question 10: How Do You Interpret the Coefficients in a Simple Linear Regression Model?

Consider the model:

$$
\hat{Y}=\beta_0+\beta_1X
$$

There are two main coefficients.

### 1. Intercept (\(\beta_0\))

The intercept represents the predicted value of Y when X is equal to 0.

Example:

$$
\hat{Y}=50+10X
$$

Here:

$$
\beta_0=50
$$

So when X = 0, predicted Y = 50.

However, the intercept is only substantively meaningful if X = 0 is within a sensible range of the data.

---

### 2. Slope (\(\beta_1\))

The slope represents the expected change in predicted Y for a **one-unit increase in X**.

In:

$$
\hat{Y}=50+10X
$$

the slope is:

$$
\beta_1=10
$$

Therefore, a one-unit increase in X is associated with a **10-unit increase in predicted Y**.

### Positive Slope

$$
\beta_1>0
$$

As X increases, predicted Y increases.

### Negative Slope

$$
\beta_1<0
$$

As X increases, predicted Y decreases.

### Zero Slope

$$
\beta_1=0
$$

There is no linear change in predicted Y as X changes.

---
