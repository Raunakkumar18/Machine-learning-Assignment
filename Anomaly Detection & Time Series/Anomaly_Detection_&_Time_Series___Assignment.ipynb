# QUESTION 1: What is Anomaly Detection? Explain its types (point, contextual, and collective anomalies) with examples.

## Answer:

**Anomaly Detection** is the process of identifying data points or patterns that significantly differ from normal behavior.

Anomalies are also called **outliers**. They may represent errors, fraud, equipment failures, unusual events, or important changes in a system.

There are three major types of anomalies:

### 1. Point Anomaly

A **point anomaly** occurs when a single observation is significantly different from the rest of the data.

### Example:

Suppose the normal daily electricity consumption of a house is between 10 and 20 kWh, but one day it is 100 kWh.

```text
12, 15, 17, 14, 16, 100, 18, 15
                  ↑
             Point Anomaly
```

The value 100 is an unusual individual observation.

---

### 2. Contextual Anomaly

A **contextual anomaly** is a data point that is abnormal only in a particular context.

The context may include:

* Time
* Location
* Season
* Weather
* User behavior

### Example:

Electricity consumption of 80 kWh may be normal for a factory during working hours but abnormal for the same factory at midnight.

Therefore, the same value can be normal in one context and anomalous in another.

---

### 3. Collective Anomaly

A **collective anomaly** occurs when a group or sequence of observations is unusual, even though individual observations may not appear abnormal.

### Example:

Suppose a server normally receives:

```text
100, 105, 110, 98, 103 requests/minute
```

Suddenly it receives:

```text
150, 155, 160, 158, 162
```

Each value might not be an extreme outlier individually, but the sequence as a whole may indicate a cyberattack or abnormal traffic pattern.

### Comparison:

| Type               | Meaning                       | Example                          |
| ------------------ | ----------------------------- | -------------------------------- |
| Point Anomaly      | One unusual observation       | ₹10,00,000 transaction           |
| Contextual Anomaly | Unusual in a specific context | High electricity use at midnight |
| Collective Anomaly | Group/sequence is unusual     | Sudden sustained server traffic  |

### Real-World Applications:

* Fraud detection
* Network intrusion detection
* Credit card monitoring
* Healthcare monitoring
* Equipment failure detection
* Energy consumption monitoring
* Cybersecurity

### Easy Definition:

**Anomaly Detection identifies observations or patterns that are significantly different from normal behavior.**

# QUESTION 2: Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of their approach and suitable use cases.

## Answer:

Isolation Forest, DBSCAN, and Local Outlier Factor (LOF) are commonly used for detecting unusual observations, but they use different approaches.

### 1. Isolation Forest

Isolation Forest detects anomalies by randomly selecting features and split values to isolate observations.

Anomalies are generally easier to isolate because they are rare and different from normal observations.

### Suitable Use Cases:

* Fraud detection
* Network attacks
* Large datasets
* Sensor anomaly detection
* Financial transactions

---

### 2. DBSCAN

DBSCAN is a density-based clustering algorithm.

It identifies dense groups of observations as clusters and points that do not belong to sufficiently dense regions can be classified as noise.

In Scikit-learn, noise points receive the label `-1`.

### Suitable Use Cases:

* Spatial data
* Geographical analysis
* Irregularly shaped clusters
* Noise detection
* Location-based data

---

### 3. Local Outlier Factor (LOF)

LOF detects anomalies by comparing the **local density** of a point with the local density of its neighbors.

If a point has substantially lower local density than its neighbors, it can be considered an outlier.

### Suitable Use Cases:

* Local anomaly detection
* Customer behavior analysis
* Sensor data
* Data with regions having different densities

### Comparison Table:

| Feature                    | Isolation Forest                | DBSCAN                       | LOF                            |
| -------------------------- | ------------------------------- | ---------------------------- | ------------------------------ |
| Main idea                  | Isolates unusual points         | Finds dense regions          | Compares local density         |
| Type                       | Anomaly detection               | Clustering + noise detection | Local outlier detection        |
| Needs number of clusters?  | No                              | No                           | No                             |
| Detects noise?             | Yes                             | Yes                          | Yes                            |
| Good for large datasets    | Yes                             | Can be                       | Moderate                       |
| Handles irregular clusters | Not its main purpose            | Yes                          | Can detect local outliers      |
| Sensitive to density       | Less directly                   | Yes                          | Yes                            |
| Main parameters            | `n_estimators`, `contamination` | `eps`, `min_samples`         | `n_neighbors`, `contamination` |

### Easy Definition:

**Isolation Forest isolates anomalies, DBSCAN finds low-density noise, and LOF compares the local density of each point with its neighbors.**

# QUESTION 3: What are the key components of a Time Series? Explain each with one example.

## Answer:

A **Time Series** is a sequence of observations recorded in chronological order.

The main components of a time series are:

### 1. Trend

Trend represents the long-term upward or downward movement of the data.

### Example:

A company's yearly sales increase continuously:

```text
2021 → ₹10 lakh
2022 → ₹12 lakh
2023 → ₹15 lakh
2024 → ₹18 lakh
```

This represents an upward trend.

---

### 2. Seasonality

Seasonality represents a pattern that repeats at regular and known intervals.

### Example:

Ice cream sales increase every summer and decrease during winter.

```text
Summer → High
Winter → Low
Summer → High
Winter → Low
```

---

### 3. Cyclical Component

A cycle represents longer-term fluctuations that do not necessarily occur at a fixed, known frequency.

### Example:

Economic activity may rise and fall over several years due to business cycles.

---

### 4. Irregular/Residual Component

This represents random or unexplained variations that remain after modeling systematic components.

### Example:

A sudden drop in sales because of an unexpected power outage or natural disaster.

### Decomposition:

A time series can often be represented as:

$$
Y_t = Trend_t + Seasonal_t + Cyclical_t + Residual_t
$$

For a multiplicative decomposition:

$$
Y_t = Trend_t \times Seasonal_t \times Cyclical_t \times Residual_t
$$

### Easy Definition:

**A time series can contain trend, seasonality, cycles, and random/residual variations.**

# QUESTION 5: Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.

## Answer:

### 1. AR – AutoRegressive Model

AR uses previous values of the time series to predict the current value.

An AR(p) model can be written as:

$$
Y_t = c+\phi_1Y_{t-1}+\phi_2Y_{t-2}+...+\phi_pY_{t-p}+\epsilon_t
$$

### Application:

Useful when current values depend on previous observations.

---

### 2. MA – Moving Average Model

MA uses previous forecast errors to predict the current value.

An MA(q) model is:

$$
Y_t = c+\epsilon_t+\theta_1\epsilon_{t-1}+...+\theta_q\epsilon_{t-q}
$$

### Application:

Useful when current observations are influenced by previous shocks/errors.

---

### 3. ARIMA

ARIMA stands for:

**AutoRegressive Integrated Moving Average**

It combines:

* AR → autoregressive component
* I → differencing
* MA → moving-average error component

Written as:

$$
ARIMA(p,d,q)
$$

Where:

* p = AR order
* d = number of differences
* q = MA order

### Application:

Non-seasonal time series forecasting.

---

### 4. SARIMA

SARIMA stands for:

**Seasonal ARIMA**

It adds seasonal AR, differencing, and MA components.

Written as:

$$
SARIMA(p,d,q)(P,D,Q,s)
$$

Where:

* p,d,q = non-seasonal orders
* P,D,Q = seasonal orders
* s = seasonal period

### Application:

Monthly airline passenger forecasting, where yearly seasonality exists.

---

### 5. SARIMAX

SARIMAX is SARIMA with **exogenous variables**.

Written conceptually as:

$$
SARIMAX(p,d,q)(P,D,Q,s)+X
$$

Where X represents external explanatory variables.

### Example:

Electricity demand can depend on:

* Past electricity demand
* Seasonality
* Temperature
* Humidity
* Holidays

### Comparison:

| Model   | Main Structure              | Application                       |
| ------- | --------------------------- | --------------------------------- |
| AR      | Previous values             | Autoregressive series             |
| MA      | Previous errors             | Error-driven series               |
| ARIMA   | AR + Differencing + MA      | Non-seasonal forecasting          |
| SARIMA  | ARIMA + Seasonality         | Seasonal forecasting              |
| SARIMAX | SARIMA + External Variables | Forecasting with external factors |

### Easy Definition:

**AR uses past values, MA uses past errors, ARIMA adds differencing, SARIMA adds seasonality, and SARIMAX additionally uses external variables.**

# QUESTION 7: Apply Isolation Forest on a Numerical Dataset (e.g., NYC Taxi Fare) to Detect Anomalies. Visualize the Anomalies on a 2D Scatter Plot.

## Answer:

Isolation Forest is an unsupervised anomaly detection algorithm that isolates unusual observations through random partitioning.

For a numerical taxi dataset, features such as:

* Fare amount
* Distance
* Passenger count

can be used.

### Google Colab Code:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest

# Create sample taxi-like numerical data
np.random.seed(42)

normal_data = np.column_stack([
    np.random.uniform(2, 30, 500),      # Fare
    np.random.uniform(1, 20, 500)       # Distance
])

# Add artificial anomalies
anomalies = np.array([
    [150, 2],
    [200, 3],
    [180, 1],
    [5, 60],
    [8, 80],
    [250, 5]
])

X = np.vstack([
    normal_data,
    anomalies
])

taxi_df = pd.DataFrame(
    X,
    columns=[
        "fare_amount",
        "distance"
    ]
)

# Isolation Forest
model = IsolationForest(
    contamination=0.02,
    random_state=42
)

taxi_df["anomaly"] = model.fit_predict(
    taxi_df[
        ["fare_amount", "distance"]
    ]
)

# -1 = anomaly
#  1 = normal

# Plot
plt.figure(figsize=(10, 6))

normal = taxi_df[
    taxi_df["anomaly"] == 1
]

outliers = taxi_df[
    taxi_df["anomaly"] == -1
]

plt.scatter(
    normal["fare_amount"],
    normal["distance"],
    s=25,
    label="Normal"
)

plt.scatter(
    outliers["fare_amount"],
    outliers["distance"],
    marker="x",
    s=100,
    label="Anomaly"
)

plt.xlabel("Fare Amount")
plt.ylabel("Distance")
plt.title("Isolation Forest Anomaly Detection")
plt.legend()

plt.show()

print(
    "Number of anomalies:",
    len(outliers)
)
```

### Output:

The scatter plot displays:

* Normal observations as regular points.
* Anomalies as `X` markers.

### Observation:

Isolation Forest identifies observations that are unusual compared with the majority of the numerical data.

**Note:** The code above uses synthetic taxi-like data so that it runs directly in Colab. If an actual NYC Taxi Fare CSV is available, replace the synthetic-data section with the real dataset and select appropriate numerical columns.

# QUESTION 8: Train a SARIMA Model on the Monthly Airline Passengers Dataset. Forecast the Next 12 Months and Visualize the Results.

## Answer:

SARIMA is suitable for the AirPassengers dataset because the data contains strong yearly seasonality.

For monthly data:

```text
Seasonal period = 12
```

### Google Colab Code:

```python
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX

# Load AirPassengers
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"

df = pd.read_csv(url)

# Convert date
df["Month"] = pd.to_datetime(
    df["Month"]
)

df.set_index(
    "Month",
    inplace=True
)

df.columns = ["Passengers"]

# Build SARIMA model
model = SARIMAX(
    df["Passengers"],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

# Train
model_fit = model.fit(
    disp=False
)

# Forecast next 12 months
forecast = model_fit.forecast(
    steps=12
)

# Print forecast
print("Next 12 Months Forecast:")
print(forecast)

# Plot
plt.figure(figsize=(12, 6))

plt.plot(
    df.index,
    df["Passengers"],
    label="Historical"
)

plt.plot(
    forecast.index,
    forecast,
    marker="o",
    label="Forecast"
)

plt.xlabel("Date")
plt.ylabel("Passengers")
plt.title(
    "SARIMA Forecast - Next 12 Months"
)

plt.legend()
plt.show()
```

### Expected Output:

The program prints 12 future monthly passenger forecasts and displays them on a graph.

### Observation:

The SARIMA model captures:

* Long-term trend
* Seasonal yearly pattern
* Short-term time-series dependence

and produces forecasts for the next 12 months.

# QUESTION 9: Apply Local Outlier Factor (LOF) on Any Numerical Dataset to Detect Anomalies and Visualize Them Using Matplotlib.

## Answer:

**Local Outlier Factor (LOF)** identifies observations whose local density is substantially lower than that of their neighbors.

### Google Colab Code:

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.neighbors import LocalOutlierFactor

# Generate normal data
np.random.seed(42)

normal_data = np.random.normal(
    loc=0,
    scale=1,
    size=(500, 2)
)

# Generate anomalies
outliers = np.array([
    [6, 6],
    [7, 5],
    [-6, -5],
    [-7, -6],
    [8, -7]
])

# Combine
X = np.vstack([
    normal_data,
    outliers
])

# LOF
model = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.02
)

labels = model.fit_predict(
    X
)

# -1 = anomaly
#  1 = normal

normal = X[
    labels == 1
]

anomalies = X[
    labels == -1
]

# Visualization
plt.figure(figsize=(9, 6))

plt.scatter(
    normal[:, 0],
    normal[:, 1],
    s=25,
    label="Normal"
)

plt.scatter(
    anomalies[:, 0],
    anomalies[:, 1],
    marker="x",
    s=100,
    label="Anomaly"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Local Outlier Factor Anomaly Detection")
plt.legend()

plt.show()

print(
    "Number of anomalies:",
    len(anomalies)
)
```

### Output:

The graph highlights the observations classified as anomalies.

### Observation:

LOF is useful when an observation is unusual relative to its **local neighborhood**, rather than only being globally far away from the rest of the data.

# Step 6: Example SARIMAX Code

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Prepare data
data = df.dropna().copy()

y = data[
    "energy_usage"
]

exog = data[
    [
        "temperature",
        "humidity"
    ]
]

# Train-test split
split = int(
    len(data) * 0.8
)

y_train = y.iloc[:split]
y_test = y.iloc[split:]

exog_train = exog.iloc[:split]
exog_test = exog.iloc[split:]

# SARIMAX model
model = SARIMAX(
    y_train,
    exog=exog_train,
    order=(1, 0, 1),
    seasonal_order=(1, 1, 1, 96),
    enforce_stationarity=False,
    enforce_invertibility=False
)

model_fit = model.fit(
    disp=False
)

# Forecast
forecast = model_fit.forecast(
    steps=len(y_test),
    exog=exog_test
)

print(forecast.head())
```

### Important:

For 15-minute data, `96` observations represent one day:

$$
24 \times 4 = 96
$$

The exact seasonal structure should be determined from the actual data rather than assumed blindly.

# Step 7: Evaluate Forecasting Performance

Use metrics such as:

### MAE

$$
MAE =
\frac{1}{n}
\sum |y_i-\hat{y}_i|
$$

### RMSE

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum(y_i-\hat{y}_i)^2
}
$$

### MAPE

$$
MAPE =
\frac{100}{n}
\sum
\left|
\frac{y_i-\hat{y}_i}{y_i}
\right|
$$

### Python:

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

import numpy as np

mae = mean_absolute_error(
    y_test,
    forecast
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        forecast
    )
)

print("MAE:", mae)
print("RMSE:", rmse)
```

# IMPORTANT FORMULAS

### MAE

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

### RMSE

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

### Differencing

$$
Y'_t=Y_t-Y_{t-1}
$$

### Seasonal Differencing

$$
Y'_t=Y_t-Y_{t-s}
$$

### Silhouette Score is NOT an anomaly metric

It is primarily a **clustering evaluation metric**, while Isolation Forest and LOF are anomaly-detection methods.

# CONCLUSION

Anomaly Detection and Time Series Analysis are important techniques for finding unusual behavior and forecasting future values.

**Isolation Forest**, **DBSCAN**, and **LOF** provide different approaches for identifying anomalies. Isolation Forest isolates unusual observations, DBSCAN uses density-based clustering and noise detection, while LOF compares the local density of observations.

For time-series forecasting, **ARIMA** is useful for non-seasonal series, **SARIMA** is useful when seasonality is present, and **SARIMAX** is useful when external variables such as weather also influence the target.

In a real-world power-grid system, combining real-time anomaly detection with SARIMAX-based forecasting can help detect abnormal energy consumption, improve demand planning, reduce operational risks, and support faster business decisions.
